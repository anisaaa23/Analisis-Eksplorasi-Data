```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# ==========================================
# 0. DATA 100 SAMPEL: IPK 100 MAHASISWA
# ==========================================
data = np.array([
    3.55, 3.20, 3.80, 3.45, 3.10, 3.35, 3.70, 3.50, 3.05, 3.40,
    3.60, 3.25, 3.75, 2.85, 3.15, 3.45, 3.65, 3.55, 2.95, 3.35,
    3.90, 3.20, 2.90, 3.40, 3.55, 3.25, 3.50, 3.75, 3.10, 3.25,
    3.45, 3.00, 3.60, 3.35, 3.85, 3.35, 3.50, 2.95, 3.25, 3.70,
    3.15, 3.45, 3.80, 3.05, 3.30, 3.65, 3.20, 3.55, 3.40, 2.95,
    3.75, 3.30, 3.50, 3.10, 3.35, 3.45, 3.90, 3.15, 3.60, 3.20,
    3.40, 3.65, 2.90, 3.25, 3.55, 3.70, 3.10, 3.35, 3.45, 3.80,
    3.20, 3.60, 3.05, 3.40, 3.30, 3.50, 3.70, 3.25, 3.45, 3.05,
    3.55, 3.35, 3.15, 3.85, 3.50, 3.40, 2.95, 3.65, 3.20, 3.70,
    3.35, 3.45, 3.25, 3.75, 3.55, 3.00, 3.40, 3.50, 3.20, 3.60
])

n_samp = len(data)

print("=== DAFTAR 100 DATA IPK MAHASISWA ===")
df_sampel = pd.DataFrame({
    'Mahasiswa Ke-': range(1, n_samp + 1),
    'IPK': data
})
print(df_sampel.to_string(index=False))
print("-" * 55)

# ==========================================
# 1. ESTIMASI TITIK DAN INTERVAL (Distribusi Z, Alpha = 5%)
# ==========================================
alpha = 0.05  # Tingkat kesalahan 5%[cite: 1]

# A. Estimasi Titik (Point Estimate): Rata-rata sampel[cite: 4, 11]
point_estimate = np.mean(data)

# Standar deviasi sampel dan standar error[cite: 4, 9]
std_dev = np.std(data, ddof=1)
std_error = std_dev / np.sqrt(n_samp)

# Nilai Z untuk tingkat kepercayaan 95% (tingkat kesalahan alpha = 5%)[cite: 9, 11]
z_score = stats.norm.ppf(1 - alpha / 2)

# B. Estimasi Interval (Confidence Interval)[cite: 7, 9]
margin_of_error = z_score * std_error
lower_bound = point_estimate - margin_of_error
upper_bound = point_estimate + margin_of_error

print("=== 1. ESTIMASI TITIK & INTERVAL (Distribusi Z) ===")
print(f"Jumlah Sampel (n)              : {n_samp}")
print(f"Estimasi Titik (Rata-rata)     : {point_estimate:.4f}")
print(f"Standar Deviasi Sampel (s)     : {std_dev:.4f}")
print(f"Tingkat Kesalahan (Alpha)      : {alpha * 100}%")
print(f"Nilai Kritis Z (Z_alpha/2)     : {z_score:.4f}")
print(f"Margin of Error                : {margin_of_error:.4f}")
print(f"Estimasi Interval (95% CI)     : [{lower_bound:.4f}, {upper_bound:.4f}]")
print("-" * 55)

# ==========================================
# 2. RESAMPLING: METODE JACKKNIFE
# ==========================================
jackknife_estimates = np.zeros(n_samp)

for i in range(n_samp):
    jack_sample = np.delete(data, i)
    jackknife_estimates[i] = np.mean(jack_sample)

mean_jackknife = np.mean(jackknife_estimates)

print("=== 2A. RESAMPLING METODE JACKKNIFE ===")
print(f"Jumlah Sub-sampel Jackknife    : {n_samp}")
print(f"Rata-rata Estimasi Jackknife   : {mean_jackknife:.4f}")
print("-" * 55)

# ==========================================
# 3. RESAMPLING: METODE BOOTSTRAP (1000 Iterasi)
# ==========================================
np.random.seed(42)
n_iterations = 1000  # Iterasi sebanyak 1000[cite: 1]
bootstrap_estimates = np.zeros(n_iterations)

for i in range(n_iterations):
    boot_sample = np.random.choice(data, size=n_samp, replace=True)  # Pengambilan dengan pengembalian[cite: 26, 28]
    bootstrap_estimates[i] = np.mean(boot_sample)

mean_bootstrap = np.mean(bootstrap_estimates)

print("=== 2B. RESAMPLING METODE BOOTSTRAP ===")
print(f"Jumlah Iterasi Bootstrap       : {n_iterations}")
print(f"Rata-rata Estimasi Bootstrap   : {mean_bootstrap:.4f}")
print("-" * 55)
```
